### Part 2: Data Structures

```python
ai_terms = [
    "Large Language Models",
    "Reinforcement Learning",
    "Generative AI",
    "Transformer Architecture",
    "Computer Vision",
]

print("Initial AI terms:", ai_terms)

ai_terms.extend(["Prompt Engineering", "AI Alignment"])

term_to_remove = "Reinforcement Learning"
if term_to_remove in ai_terms:
    ai_terms.remove(term_to_remove)

print("Updated AI terms:", ai_terms)
```

### Part 3: Conditional Statements

```python
user_input = input("Enter a number: ")

try:
    number = float(user_input)

    if number > 0:
        print(f"{number} is Positive")
    elif number < 0:
        print(f"{number} is Negative")
    else:
        print(f"{number} is Zero")

except ValueError:
    print(f"'{user_input}' is not a valid number. Please enter a numeric value.")
```
