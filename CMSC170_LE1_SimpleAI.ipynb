### Part 3: Conditional Statements

```python
user_input = input("Enter a number: ")

try:
    number = float(user_input)

    if number > 0:
        print(f"{number} is Positive")
    elif number < 0:
        print(f"{number} is Negative")
    else:
        print(f"{number} is Zero")

except ValueError:
    print(f"'{user_input}' is not a valid number. Please enter a numeric value.")
```

# Task 3: Simple AI Algorithm

### Part 1: Rule-based Classifier

```python
user_input_term = input("Enter a term to classify: ")

def classify_term(term):
    ai_terms = [
        "Neural Networks",
        "Machine Learning",
        "Deep Learning",
        "AI Ethics",
        "Natural Language Processing",
        "Computer Vision",
        "Robotics"
    ]

    if term in ai_terms:
        return "AI-related"
    else:
        return "Not AI-related"

print(classify_term(user_input_term))
```

### Challenge 1: Classify Multiple Terms

```python
terms = input("Enter terms separated by commas: ").split(",")

def classify_term(term):
    ai_terms = [
        "Neural Networks",
        "Machine Learning",
        "Deep Learning",
        "AI Ethics",
        "Natural Language Processing",
        "Computer Vision",
        "Robotics"
    ]

    if term in ai_terms:
        return "AI-related"
    else:
        return "Not AI-related"

for term in terms:
    clean = term.strip()
    print(f"{clean}: {classify_term(clean)}")
```

### Challenge 2: Rule-based Chatbot

```python
def chatbot_response(user_input):
    text = user_input.lower().strip()

    if any(word in text.split() for word in ["hello", "hi", "hey"]):
        return "Hi there! How can I help you today?"

    elif "machine learning" in text:
        return "Machine Learning is a subset of AI where systems learn from data."

    elif "ai" in text or "artificial intelligence" in text:
        return "AI is the simulation of human intelligence in machines."

    elif "your name" in text or "who are you" in text:
        return "I'm a simple rule-based chatbot!"

    elif "how are you" in text:
        return "I'm just code, but I'm running smoothly! How about you?"

    elif "help" in text:
        return "You can ask me about AI, machine learning, or just say hello!"

    else:
        return "Hmm, I'm not sure I understand. Could you rephrase that, or ask me about AI?"


def run_chatbot():
    print("Chatbot: Hi! I'm a rule-based chatbot. Type 'bye' to exit.")

    while True:
        user_input = input("You: ")
        print(f"User: {user_input}")

        if user_input.strip() == "":
            print("Chatbot: Please type something!")
            continue
        if user_input.lower().strip() == "bye":
            print("Chatbot: Goodbye!")
            break

        response = chatbot_response(user_input)
        print(f"Chatbot: {response}")


run_chatbot()
```
